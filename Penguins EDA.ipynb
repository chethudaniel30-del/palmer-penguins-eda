```python
plt.figure(figsize=(8, 6))

numeric_cols = [
    'bill_length_mm',
    'bill_depth_mm',
    'flipper_length_mm',
    'body_mass_g'
]

correlation_matrix = df[numeric_cols].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Between Penguin Body Measurements')

plt.savefig(
    'visualization_6_correlation_heatmap.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
from docx import Document
from docx.shared import Inches, Pt, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_CELL_VERTICAL_ALIGNMENT
from docx.oxml import OxmlElement
from docx.oxml.ns import qn
from docx.enum.section import WD_SECTION
import os

# --------------------------------------------------
# 1. CREATE DOCUMENT
# --------------------------------------------------

doc = Document()

section = doc.sections[0]
section.top_margin = Inches(0.7)
section.bottom_margin = Inches(0.7)
section.left_margin = Inches(0.8)
section.right_margin = Inches(0.8)

# Default font
styles = doc.styles
styles['Normal'].font.name = 'Calibri'
styles['Normal'].font.size = Pt(10.5)

styles['Title'].font.name = 'Calibri'
styles['Title'].font.size = Pt(26)
styles['Title'].font.bold = True
styles['Title'].font.color.rgb = RGBColor(31, 78, 121)

for heading in ['Heading 1', 'Heading 2']:
    styles[heading].font.name = 'Calibri'
    styles[heading].font.color.rgb = RGBColor(31, 78, 121)

# Helper functions
def add_heading(text, level=1):
    doc.add_heading(text, level=level)

def add_paragraph(text):
    doc.add_paragraph(text)

def add_code(code):
    p = doc.add_paragraph()
    p.paragraph_format.left_indent = Inches(0.2)
    p.paragraph_format.space_after = Pt(8)
    run = p.add_run(code)
    run.font.name = 'Consolas'
    run.font.size = Pt(8)
    run.font.color.rgb = RGBColor(60, 60, 60)

def add_table(headers, rows):
    table = doc.add_table(rows=1, cols=len(headers))
    table.style = 'Light Shading Accent 1'
    table.alignment = WD_TABLE_ALIGNMENT.CENTER

    for i, header in enumerate(headers):
        table.rows[0].cells[i].text = str(header)

    for row in rows:
        cells = table.add_row().cells
        for i, value in enumerate(row):
            cells[i].text = str(value)

    for row in table.rows:
        for cell in row.cells:
            cell.vertical_alignment = WD_CELL_VERTICAL_ALIGNMENT.CENTER

    doc.add_paragraph()

def add_visualization(number, title, filename, interpretation, code):
    add_heading(f"Visualization {number}: {title}", 2)

    image_path = os.path.join(os.getcwd(), filename)

    if os.path.exists(image_path):
        p = doc.add_paragraph()
        p.alignment = WD_ALIGN_PARAGRAPH.CENTER
        p.add_run().add_picture(image_path, width=Inches(5.8))
    else:
        add_paragraph(f"Image file not found: {filename}")

    add_paragraph("Interpretation: " + interpretation)

    p = doc.add_paragraph()
    r = p.add_run("Python code:")
    r.bold = True

    add_code(code)
    doc.add_paragraph()

# --------------------------------------------------
# 2. TITLE PAGE
# --------------------------------------------------

title = doc.add_paragraph()
title.alignment = WD_ALIGN_PARAGRAPH.CENTER
title.paragraph_format.space_before = Inches(1.5)

run = title.add_run("EXPLORATORY DATA ANALYSIS\nAND DATA VISUALIZATION")
run.bold = True
run.font.size = Pt(25)
run.font.color.rgb = RGBColor(31, 78, 121)

subtitle = doc.add_paragraph()
subtitle.alignment = WD_ALIGN_PARAGRAPH.CENTER
subtitle.paragraph_format.space_before = Inches(0.4)

run = subtitle.add_run("A Study of the Palmer Penguins Dataset")
run.font.size = Pt(17)
run.italic = True

doc.add_paragraph("\n\n")

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
p.add_run("WEEK 2 PROJECT REPORT").bold = True

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
p.add_run("Virtual Data Science with Python Internship")

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
p.add_run("Tools: Python | Pandas | NumPy | Matplotlib | Seaborn")

doc.add_page_break()

# --------------------------------------------------
# 3. INTRODUCTION
# --------------------------------------------------

add_heading("1. Introduction")

add_paragraph(
    "Exploratory Data Analysis (EDA) is an important stage of data science "
    "that involves examining datasets to understand their structure, "
    "identify patterns, discover relationships, and detect possible "
    "anomalies. Data visualization helps communicate these findings "
    "through graphical representations."
)

add_paragraph(
    "This project uses the Palmer Penguins dataset to explore physical "
    "measurements and species distributions among Adelie, Chinstrap, "
    "and Gentoo penguins. The analysis uses Python libraries to perform "
    "data inspection, statistical analysis, visualization, and comparison."
)

add_heading("2. Objectives")

for item in [
    "Understand the structure and characteristics of a public dataset.",
    "Perform initial data inspection and descriptive statistical analysis.",
    "Identify missing values and examine data quality.",
    "Create visualizations to explore distributions and relationships.",
    "Compare physical measurements across penguin species and islands.",
    "Interpret the findings and discuss limitations."
]:
    doc.add_paragraph(item, style='List Bullet')

# --------------------------------------------------
# 4. DATASET DESCRIPTION
# --------------------------------------------------

add_heading("3. Dataset Description")

add_paragraph(
    "The Palmer Penguins dataset contains observations of three penguin "
    "species from islands in the Palmer Archipelago, Antarctica. It "
    "includes categorical information and physical measurements."
)

add_paragraph(
    "Dataset source: "
    "https://github.com/mcnakhaee/palmerpenguins/blob/master/"
    "palmerpenguins/data/penguins.csv"
)

add_paragraph(
    "The dataset contains 344 observations and 8 columns. The columns "
    "include species, island, bill length, bill depth, flipper length, "
    "body mass, sex, and year."
)

add_table(
    ["Column", "Description"],
    [
        ["species", "Penguin species"],
        ["island", "Island where the penguin was observed"],
        ["bill_length_mm", "Bill length in millimetres"],
        ["bill_depth_mm", "Bill depth in millimetres"],
        ["flipper_length_mm", "Flipper length in millimetres"],
        ["body_mass_g", "Body mass in grams"],
        ["sex", "Recorded sex of the penguin"],
        ["year", "Year of observation"]
    ]
)

# --------------------------------------------------
# 5. TOOLS
# --------------------------------------------------

add_heading("4. Tools and Libraries Used")

add_table(
    ["Library", "Purpose"],
    [
        ["Pandas", "Data loading, inspection, grouping and analysis"],
        ["NumPy", "Numerical operations"],
        ["Matplotlib", "Plot creation and customization"],
        ["Seaborn", "Statistical data visualization"],
        ["python-docx", "Generation of the final Word report"]
    ]
)

add_paragraph(
    "The CSV dataset was loaded into a Pandas DataFrame. "
    "Matplotlib and Seaborn were used to generate graphs, while "
    "Pandas was used for descriptive statistics, missing-value "
    "checks, correlation analysis, and group comparisons."
)

add_code(
"""import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("penguins.csv")"""
)

# --------------------------------------------------
# 6. DATA OVERVIEW
# --------------------------------------------------

add_heading("5. Initial Data Inspection")

add_paragraph(
    "The dataset was examined using head(), shape, info(), and "
    "describe() to understand its structure, data types, and "
    "statistical properties."
)

add_paragraph("Dataset dimensions: 344 rows and 8 columns.")

add_table(
    ["Column", "Data type"],
    [
        ["species", "Object"],
        ["island", "Object"],
        ["bill_length_mm", "Float64"],
        ["bill_depth_mm", "Float64"],
        ["flipper_length_mm", "Float64"],
        ["body_mass_g", "Float64"],
        ["sex", "Object"],
        ["year", "Integer"]
    ]
)

add_heading("5.1 Descriptive Statistics", 2)

add_table(
    ["Measurement", "Mean", "Median", "Minimum", "Maximum"],
    [
        ["Bill length (mm)", "43.92", "44.45", "32.1", "59.6"],
        ["Bill depth (mm)", "17.15", "17.3", "13.1", "21.5"],
        ["Flipper length (mm)", "200.92", "197", "172", "231"],
        ["Body mass (g)", "4201.75", "4050", "2700", "6300"]
    ]
)

add_paragraph(
    "The descriptive statistics provide an overview of the central "
    "tendency and spread of the numerical measurements. The four "
    "measurement columns each contain 342 valid observations because "
    "two values are missing in each column."
)

# --------------------------------------------------
# 7. MISSING VALUES
# --------------------------------------------------

add_heading("6. Missing-Value Analysis")

add_table(
    ["Column", "Missing values"],
    [
        ["species", "0"],
        ["island", "0"],
        ["bill_length_mm", "2"],
        ["bill_depth_mm", "2"],
        ["flipper_length_mm", "2"],
        ["body_mass_g", "2"],
        ["sex", "11"],
        ["year", "0"],
        ["Total", "19"]
    ]
)

add_paragraph(
    "The dataset contains 19 missing cells. The four numerical "
    "measurement columns contain two missing values each, while "
    "the sex column contains 11 missing values. The species, island, "
    "and year columns have no missing values."
)

add_paragraph(
    "The missing values were retained rather than automatically "
    "deleting entire rows. This allowed available observations to "
    "be used in analyses wherever possible. Individual calculations "
    "may therefore use slightly different sample sizes."
)

add_code(
"""missing_values = df.isnull().sum()
print(missing_values)
print(df.isnull().sum().sum())"""
)

# --------------------------------------------------
# 8. VISUALIZATIONS
# --------------------------------------------------

add_heading("7. Exploratory Data Analysis and Visualizations")

add_paragraph(
    "Eight visualizations were created to examine species counts, "
    "measurement distributions, differences between species, "
    "correlations, and species distribution across islands."
)

add_visualization(
    1,
    "Distribution of Penguin Species",
    "visualization_1_species_distribution.png",
    "The dataset contains 152 Adelie penguins, 124 Gentoo penguins, "
    "and 68 Chinstrap penguins. Adelie is the most frequently "
    "represented species, while Chinstrap is the least represented.",
    """sns.countplot(data=df, x='species')
plt.title('Distribution of Penguin Species')
plt.xlabel('Penguin Species')
plt.ylabel('Number of Observations')
plt.show()"""
)

add_visualization(
    2,
    "Distribution of Penguin Body Mass",
    "visualization_2_body_mass_distribution.png",
    "The histogram shows variation in penguin body mass, with "
    "multiple peaks rather than one single concentrated distribution. "
    "This pattern may reflect differences in body mass among species.",
    """sns.histplot(data=df, x='body_mass_g', bins=20, kde=True)
plt.title('Distribution of Penguin Body Mass')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.show()"""
)

add_visualization(
    3,
    "Body Mass Distribution by Species",
    "visualization_3_body_mass_by_species.png",
    "Gentoo penguins generally have higher body mass values, with "
    "most observations concentrated around 4700–5500 g. Adelie "
    "and Chinstrap penguins have lower and partially overlapping "
    "distributions, mainly around 3400–4000 g. Species differences "
    "help explain the multiple peaks in the overall histogram.",
    """sns.histplot(
    data=df, x='body_mass_g', hue='species',
    bins=20, kde=True, element='step'
)
plt.title('Body Mass Distribution by Penguin Species')
plt.show()"""
)

add_visualization(
    4,
    "Flipper Length by Species",
    "visualization_4_flipper_length.png",
    "Gentoo penguins have the highest median flipper length at "
    "approximately 216 mm, followed by Chinstrap at 196 mm and "
    "Adelie at 190 mm. Adelie penguins also show potential outliers. "
    "Flipper length is therefore a useful feature for comparing "
    "the species.",
    """sns.boxplot(data=df, x='species', y='flipper_length_mm')
plt.title('Flipper Length Distribution by Penguin Species')
plt.xlabel('Penguin Species')
plt.ylabel('Flipper Length (mm)')
plt.show()"""
)

add_visualization(
    5,
    "Relationship Between Flipper Length and Body Mass",
    "visualization_5_flipper_vs_mass.png",
    "The scatter plot shows a positive relationship between flipper "
    "length and body mass. Penguins with longer flippers generally "
    "tend to have higher body mass. Species groupings are also visible, "
    "suggesting that differences between species may contribute "
    "to this overall relationship.",
    """sns.scatterplot(
    data=df, x='flipper_length_mm',
    y='body_mass_g', hue='species', s=70
)
plt.title('Relationship Between Flipper Length and Body Mass')
plt.legend(title='Penguin Species')
plt.show()"""
)

add_visualization(
    6,
    "Correlation Between Penguin Body Measurements",
    "visualization_6_correlation_heatmap.png",
    "The correlation heatmap summarizes the relationships among "
    "bill length, bill depth, flipper length, and body mass. The "
    "previously calculated correlation between flipper length and "
    "body mass is approximately 0.8712, indicating a strong positive "
    "linear relationship. Correlation does not establish causation.",
    """numeric_cols = [
    'bill_length_mm', 'bill_depth_mm',
    'flipper_length_mm', 'body_mass_g'
]
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(
    correlation_matrix, annot=True,
    cmap='coolwarm', fmt='.2f',
    vmin=-1, vmax=1
)
plt.title('Correlation Between Penguin Body Measurements')
plt.show()"""
)

add_visualization(
    7,
    "Bill Length Distribution by Species",
    "visualization_7_bill_length_by_species.png",
    "Adelie penguins generally have shorter bill lengths, with "
    "whiskers extending approximately from 35 mm to 46 mm. Gentoo "
    "penguins extend approximately from 40 mm to 56 mm, while "
    "Chinstrap penguins extend approximately from 40 mm to 58 mm. "
    "The overlapping distributions indicate that bill length alone "
    "may not reliably distinguish every individual penguin.",
    """sns.boxplot(data=df, x='species', y='bill_length_mm')
plt.title('Bill Length Distribution by Penguin Species')
plt.xlabel('Penguin Species')
plt.ylabel('Bill Length (mm)')
plt.show()"""
)

add_visualization(
    8,
    "Penguin Species Distribution Across Islands",
    "visualization_8_species_by_island.png",
    "Adelie penguins appear on all three islands. Gentoo penguins "
    "appear on Biscoe Island, while Chinstrap penguins appear on "
    "Dream Island. Biscoe has the largest overall number of "
    "observations in this dataset.",
    """sns.countplot(
    data=df, x='island', hue='species'
)
plt.title('Penguin Species Distribution Across Islands')
plt.xlabel('Island')
plt.ylabel('Number of Penguins')
plt.legend(title='Species')
plt.show()"""
)

# --------------------------------------------------
# 9. SPECIES-WISE ANALYSIS
# --------------------------------------------------

add_heading("8. Species-wise Numerical Analysis")

add_paragraph(
    "The mean values of the four physical measurements were "
    "calculated separately for each species using Pandas groupby()."
)

add_table(
    ["Measurement", "Adelie", "Chinstrap", "Gentoo"],
    [
        ["Bill length (mm)", "38.79", "48.83", "47.50"],
        ["Bill depth (mm)", "18.35", "18.42", "14.98"],
        ["Flipper length (mm)", "189.95", "195.82", "217.19"],
        ["Body mass (g)", "3700.66", "3733.09", "5076.02"]
    ]
)

add_paragraph(
    "Adelie penguins have the shortest average bill length and "
    "flipper length. Chinstrap penguins have the highest average "
    "bill length and bill depth. Gentoo penguins have the greatest "
    "average flipper length and body mass. Adelie and Chinstrap "
    "have relatively similar average body masses."
)

add_code(
"""species_analysis = df.groupby('species')[
    ['bill_length_mm', 'bill_depth_mm',
     'flipper_length_mm', 'body_mass_g']
].mean()

print(species_analysis.round(2))"""
)

# --------------------------------------------------
# 10. ISLAND-WISE ANALYSIS
# --------------------------------------------------

add_heading("9. Island-wise Body Mass Analysis")

add_table(
    ["Island", "Average body mass (g)"],
    [
        ["Biscoe", "4716.02"],
        ["Dream", "3712.90"],
        ["Torgersen", "3706.37"]
    ]
)

add_paragraph(
    "Biscoe Island has the highest average body mass at approximately "
    "4716.02 g. Dream and Torgersen have lower and relatively similar "
    "averages. This difference may partly reflect species composition, "
    "because Gentoo penguins, which have a higher average body mass, "
    "occur on Biscoe Island in this dataset. The result does not "
    "establish that island location itself causes differences in mass."
)

add_code(
"""island_analysis = df.groupby('island')['body_mass_g'].mean()
print(island_analysis.round(2))"""
)

# --------------------------------------------------
# 11. CHALLENGES AND REFLECTION
# --------------------------------------------------

add_heading("10. Challenges and Reflective Discussion")

add_paragraph(
    "One challenge during the project was setting up and maintaining "
    "the Python notebook environment. The notebook was launched "
    "through Python's notebook module when the direct command was "
    "not recognized. A kernel restart also cleared the variables "
    "stored in memory, resulting in a NameError when an earlier "
    "analysis cell was executed. Re-running the import and dataset "
    "loading cells restored the working environment."
)

add_paragraph(
    "Another challenge was handling missing values. The dataset "
    "contained missing measurements and missing sex entries. "
    "Rather than deleting observations without considering their "
    "effect, the missing-value distribution was inspected and "
    "available data were retained for the analysis."
)

add_paragraph(
    "Interpreting the visualizations also required care. For example, "
    "the overall body mass histogram contained multiple peaks. "
    "Comparing body mass by species helped show that the peaks were "
    "associated with differences between species. Similarly, "
    "island-wise body mass averages were interpreted alongside "
    "species distribution to avoid attributing differences to "
    "location alone."
)

add_paragraph(
    "The project demonstrated the importance of combining graphs "
    "with numerical analysis. Visualizations help identify patterns, "
    "while statistical summaries provide more precise comparisons. "
    "It also highlighted the need to distinguish correlation from "
    "causation and to avoid making conclusions beyond what the "
    "dataset supports."
)

# --------------------------------------------------
# 12. LIMITATIONS
# --------------------------------------------------

add_heading("11. Limitations")

for item in [
    "The dataset contains missing values, which reduce the number of valid observations for some calculations.",
    "The numbers of observations are not equal across species and islands.",
    "The analysis is observational and cannot establish causal relationships.",
    "Some visual interpretations are approximate and depend on the graph scale and grouping.",
    "The findings describe this dataset and should not automatically be generalized to every penguin population."
]:
    doc.add_paragraph(item, style='List Bullet')

# --------------------------------------------------
# 13. CONCLUSION
# --------------------------------------------------

add_heading("12. Conclusion")

add_paragraph(
    "This project used Exploratory Data Analysis and visualization "
    "techniques to examine the Palmer Penguins dataset. The analysis "
    "identified differences in physical measurements among Adelie, "
    "Chinstrap, and Gentoo penguins. Gentoo penguins showed the "
    "highest average flipper length and body mass, while Chinstrap "
    "penguins had the highest average bill length and bill depth."
)

add_paragraph(
    "A strong positive correlation of approximately 0.8712 was "
    "observed between flipper length and body mass. The island-wise "
    "analysis also showed differences in average body mass, which "
    "may partly be explained by the distribution of species across "
    "islands."
)

add_paragraph(
    "Overall, the project provided practical experience in data "
    "inspection, missing-value analysis, statistical summaries, "
    "visualization, correlation analysis, and interpretation using "
    "Python. It demonstrated how combining numerical and graphical "
    "methods can provide a clearer understanding of a dataset."
)

# --------------------------------------------------
# 14. SAVE DOCUMENT
# --------------------------------------------------

output_file = "Penguins_EDA_Visualization_Report.docx"

doc.save(output_file)

print("REPORT CREATED SUCCESSFULLY!")
print("Saved at:", os.path.abspath(output_file))
print("File size:", round(os.path.getsize(output_file) / 1024, 2), "KB")
```
